```python
import re

test_text = """
Python developer with experience in C++, C#, .NET, Node.js,
JavaScript, 3D modeling, 1C and Machine Learning.
"""

tokens = re.findall(
    r"(?u)(?:[a-zA-Z]+\+\+|[a-zA-Z]+#|\.[a-zA-Z]+|[a-zA-Z]+\.[a-zA-Z]+|\w+)",
    test_text.lower()
)

tokens
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

def technical_tokenizer(text):
    return re.findall(
        r"(?u)(?:[a-zA-Z]+\+\+|[a-zA-Z]+#|\.[a-zA-Z]+|[a-zA-Z]+\.[a-zA-Z]+|\w+)",
        text.lower()
    )

tfidf_final = TfidfVectorizer(
    tokenizer=technical_tokenizer,
    token_pattern=None,
    max_features=5000,
    ngram_range=(1, 2),
    min_df=5,
    stop_words="english"
)

X_tfidf_final = tfidf_final.fit_transform(
    model_df["Candidate_Text_NLP"].head(5000)
)

print(X_tfidf_final.shape)
```

```python
features_final = tfidf_final.get_feature_names_out()

{
    term: term in features_final
    for term in [
        "python",
        "javascript",
        "java",
        "android",
        "3d",
        "1c",
        "computer vision",
        "machine learning",
        "data science",
        "azure",
        "c++",
        "c#",
        ".net",
        "node.js"
    ]
}
```
